## 1. Warm-up: Atherogenic Index 

The **Atherogenic Index** assesses cardiovascular disease risk using the LDL/HDL ratio.

**Formula:** 
$$\text{Atherogenic Index} = \frac{\text{LDL}}{\text{HDL}}$$

**Interpretation:**
| Index | Risk Level |
|-------|------------|
| < 2.0 | No risk |
| 2.0 – 4.0 | Moderate risk |
| > 4.0 | High risk |

Here's a working implementation. Study it carefully:

````python
def calculate_atherogenic_index(ldl, hdl):
    """
    Calculate the Atherogenic Index and return risk category.
    
    Args:
        ldl: LDL cholesterol (mg/dL)
        hdl: HDL cholesterol (mg/dL)
    
    Returns:
        Tuple of (index_value, risk_category)
    """
    index = ldl / hdl
    
    if index < 2.0:
        risk = "No risk"
    elif index <= 4.0:  # Note: 2.0-4.0 inclusive
        risk = "Moderate risk"
    else:
        risk = "High risk"
    
    return index, risk


# Test with two patients
index1, risk1 = calculate_atherogenic_index(ldl=4, hdl=3.5)
print(f"Patient 1: Index = {index1:.2f} → {risk1}")

index2, risk2 = calculate_atherogenic_index(ldl=12, hdl=2.5)
print(f"Patient 2: Index = {index2:.2f} → {risk2}")
````

### Adding Input Validation

Clinical data can have errors. Let's add validation:

````python
def calculate_atherogenic_index_safe(ldl, hdl):
    """
    Calculate Atherogenic Index with input validation.
    """
    # Validate inputs
    if hdl <= 0:
        return None, "Error: HDL must be positive"
    if ldl <= 0:
        return None, "Error: LDL must be positive"
    
    # Calculate
    index = ldl / hdl
    
    if index < 2.0:
        risk = "No risk"
    elif index <= 4.0:
        risk = "Moderate risk"
    else:
        risk = "High risk"
    
    return index, risk


# Test with invalid input
result = calculate_atherogenic_index_safe(ldl=100, hdl=0)
print(f"With HDL=0: {result}")
````
