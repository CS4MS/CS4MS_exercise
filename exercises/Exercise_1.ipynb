## 0. Before We Write Code: Understanding the Setup

Before touching any Python, let's orient ourselves. Most of you have used software with buttons, menus, and forms: tools like Excel or hospital information systems. Programming feels different at first, but it follows the same logic. You're just giving instructions in a more explicit way.

This section answers the questions students most often have (but rarely ask):
- *What even is a coding environment?*

So let's explore this!


### What Is a Programming Environment?

Think of it like this: when a surgeon operates, they need a sterile room, the right instruments, and a trained team. The **operating room** is the environment — it sets up the conditions so the actual work can happen.

A **coding environment** does the same thing for code. It bundles together:

| Component | What it is | Analogy |
|-----------|-----------|--------|
| **Editor / IDE** | Where you write code | The place where you write your surgical notes |
| **Python interpreter** | What reads and runs your code | A colleague who executes your instructions |
| **File system** | Where your data/files are stored | The hospital's record archive |
| **Libraries** | Pre-built tools you can import | Specialized equipment you wheel in when needed |

Popular environments you might encounter:
- **Google Colab** ← what we use here; runs entirely in your browser, nothing to install
- **VS Code** — a powerful desktop editor used by professional developers
- **JupyterLab** — similar to Colab, but runs on your own computer

> 💡 **Key insight:** The environment doesn't write code for you — it just makes sure Python is available and your files are accessible. The logic is always yours.


### What Is This Notebook?

You're looking at a **Jupyter Notebook** (running in Google Colab). It's a document that mixes two things:

- **Markdown cells** — formatted text, like this one. Think of these as your lab notes or explanations.
- **Code cells** — the grey boxes with `[ ]` on the left. These contain Python code you can actually run. See the one below: 


````python
print("This is a code cell!")
````

**How to run a code cell:** Click inside it and press **Shift + Enter** (or click the ▶ button on the left). The output appears directly below.

**The `[ ]` counter:** When a cell runs, the `[ ]` fills in with a number like `[1]`. This tells you the **order** cells were executed — important because cells can affect each other.

> ⚠️ **Common confusion:** Cells in a notebook share memory, but only if you've actually run them. If you define a variable in Cell 3 but skip Cell 2, Cell 3 might fail because something from Cell 2 is missing. Always run cells top to bottom.

### Where Does My Code Actually Run?

When you run a cell in Google Colab, your code doesn't run on your laptop — it runs on a **virtual machine** (a computer) in Google's data centers. This is why:

- You don't need to install Python
- A slow laptop doesn't matter much
- But: if you close the tab and come back, **variables are gone** (you have to re-run cells)

```
Your Browser  ──sends code──►  Google's Computer  ──returns──►  Output in your notebook
   (Colab)                      (runs Python)                    (text, numbers, plots)
```

On your own computer, the same flow happens locally — Python is installed, and the "computer" is just your own machine.


### Files and the File System

One of the most confusing things for new programmers: **where are my files?**

Your computer organizes files in a tree of folders. Python sees this tree too, and can read, write, and move files just like you can in a file browser — just using code instead of clicking.

```
📁 My Computer
   📁 Documents
      📁 Medical_Course
         📄 patient_data.csv      ← a data file
         📄 exercise_1.ipynb      ← this notebook
         📁 results
            📄 analysis.png       ← a plot we might save
```

In Python, you refer to files by their **path** — the address within this tree:
- `"Documents/Medical_Course/patient_data.csv"` on your own computer
- `"/content/patient_data.csv"` in Google Colab (Colab has its own temporary file system)

You can actually ask Python what files it can see right now:


````python
# Python can talk to the file system using the built-in 'os' module
# 'os' stands for Operating System — it lets Python ask the OS questions
import os

# Ask: what folder are we currently in?
current_folder = os.getcwd()  # getcwd = get current working directory
print(f"We are currently in: {current_folder}")
````

> 🩺 **The bottom line:** Python is a language your computer understands. A notebook is a document where you write that language, run it, and see the results all in one place. Files are just data sitting on disk; Python can reach them when you point it to the right path.

## 2. Control Flow Statements 

### If-Else: Making Decisions

**Important:** Python uses indentation (spaces at the start of a line) to group code. The indented lines below `if:` only run when the condition is true.

````python
temperature = 38.5

if temperature > 38.0:
    print("Patient has fever")
else:
    print("Temperature normal")
````

Use `elif` (else-if) for multiple conditions:

````python
heart_rate = 95

if heart_rate < 60:
    print("Bradycardia")
elif heart_rate > 100:
    print("Tachycardia")
else:
    print("Normal heart rate")
````

### For Loops: Repeating Actions

Use `for` to repeat code a specific number of times or iterate over items.

````python
# range(5) gives: 0, 1, 2, 3, 4
for i in range(5):
    print(f"Measurement {i}")
````

````python
# Sum the first 10 numbers
total = 0
for i in range(1, 11):  # range(1, 11) gives: 1, 2, ..., 10
    total += i  # same as: total = total + i

print(f"Sum: {total}")
````

## 3. Lists and Dictionaries

### Lists: Ordered Collections

Lists store multiple items in order. Use square brackets `[]`.

````python
vitals = [120, 80, 72, 37.2]  # SBP, DBP, HR, Temp

# Access by index (starts at 0!)
print(f"Systolic BP: {vitals[0]}")
print(f"Heart rate: {vitals[2]}")

# Negative index = from the end
print(f"Last value: {vitals[-1]}")
````

````python
# Useful list functions
numbers = [4, 8, 2, 9, 1]

print(f"Length: {len(numbers)}")
print(f"Sum: {sum(numbers)}")
print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")
````

````python
# Add items with append()
medications = ["Aspirin", "Metformin"]
medications.append("Lisinopril")
print(medications)
````

````python
# Loop through a list
for med in medications:
    print(f"Patient takes: {med}")
````

### Dictionaries: Key-Value Pairs

Dictionaries store data with named keys (like a patient record). Use curly brackets `{}`.

````python
patient = {
    "name": "Maria Schmidt",
    "age": 45,
    "diagnosis": "Type 2 Diabetes",
    "admitted": True
}

# Access by key
print(f"Patient: {patient['name']}")
print(f"Age: {patient['age']}")
````

````python
# Update values
patient["age"] = 46

# Add new keys
patient["room"] = "B204"

print(patient)
````

## 4. Functions 

Functions are reusable blocks of code. Define them once, use them many times.

````python
def greet(name):
    """Print a greeting for the given name."""
    print(f"Hello, {name}!")

greet("Dr. Müller")
greet("Mr. Weber")
````

### Return Values

Use `return` to send a result back from the function.

````python
def calculate_bmi(weight_kg, height_m):
    """Calculate Body Mass Index."""
    bmi = weight_kg / (height_m ** 2)
    return bmi

# Use the function
result = calculate_bmi(70, 1.75)
print(f"BMI: {result:.1f}")
````

### Reading Error Messages

When code fails, Python shows a **traceback**. Here's how to read it:

````python
# This will cause an error - run it and read the message
def divide(a, b):
    return a / b

result = divide(10, 0)  # Can't divide by zero!
````

**How to read the error:**
1. Look at the **last line** — it tells you the error type (`ZeroDivisionError`)
2. Look at the **arrow** — it shows which line caused the error
3. Read the **message** — it explains what went wrong
